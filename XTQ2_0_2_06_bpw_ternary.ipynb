```python
د:
💡 النموذج تم تدريبه منذ البداية كـ "نموذج ثلاثي" (Native Ternary Training)
1. ما الفرق بين ما حدث هنا وما حدث مع Qwen و SmolLM؟
في النماذج العادية (مثل Qwen و SmolLM):
يتم تدريبها بأرقام عشرية كاملة الدقة (FP16/BF16) بملايين التدرجات (مثلاً: 0.3421, -1.8742, 0.0091).
عندما قمت بتطبيق تكميم التيرناري TQ2_0 عليها، أنت أجبرت هذه الملايين من التدرجات أن تصبح فقط 3 خيارات لا رابع لها: {-1، 0، +1}.
هذا أدّى إلى فقدان 90% من المعلومات المخزنة في الأوزان فتدمر النموذج وأصبح يهذي.
في نموذج TriLM الذي نجح معك الآن:
تم تدريبه في المختبر من الصفر (From Scratch) ليعمل فقط بثلاث قيم لكل وزن: {-1، 0، +1} (وهي تقنية مستوحاة من أبحاث BitNet b1.58).
شبكته العصبية تعلّمت اللغة والتفكير بالاعتماد الكامل على هذه القيم الثلاث فقط!
2. ماذا فعل أمر التكميم quantize ... TQ2_0 تحديداً مع TriLM؟
عندما قمت بتشغيل الأمر:
لم تقم الأداة "بقصّ" أو "تخمين" أو "تدمير" أي معلومات.
الأداة قامت فقط بعملية تعبئة وتغليف (Packing): الأوزان كانت أصلًا مخزنة شكلياً داخل ملف كبير (FP16) وكل وزن يأخذ 16 بت، فجاءت صيغة TQ2_0 وضغطتها لتأخذ 2 بت فقط لكل وزن (لأن 2 بت كافية رياضياً لتخزين الحالات الثلاث: -1، 0، 1).
تشبيه بسيط يوضح الفكرة:
Qwen و SmolLM: مثل لوحة زيتية ملونة فيها 16 مليون درجة لون، جئت أنت وطبعتها فجأة بحبر أبيض وأسود فقط بدون أي تدرج رمادي 👈 النتيجة: بقع سوداء غير مفهومة وضياع كامل للملامح.
TriLM: مثل رسم كاريكاتيري بالأساس رُسم بالورقة والقلم بـ 3 ألوان فقط (أبيض، رمادي، أسود) 👈 عندما طبعته بنفس الألوان الثلاثة، ظهرت كل ملامح الرسمة واضحة ومتقنة تماماً!
لهذا السبب نموذج TriLM بحجم 400 ميجابايت فقط أعطاك إجابة لغوية منطقية وسريعة جداً.
8.8s
info
Google AI models may make mistakes, so double-check outputs.
Use Arrow Up and Arrow Down to select a turn, Enter to jump to it, and Escape to return to the chat.
Start typing a prompt to see what our models can do


google
Grounding with Google Search


```
